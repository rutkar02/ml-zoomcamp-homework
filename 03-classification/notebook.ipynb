# One hot encoding

```python
# We convert dataframe into dictionaries using this
dicts = df[categorical + numerical].to_dict(orient='records')
#  The orient records argument means one dictionary per row
```

```python
dicts
```

```python
from sklearn.feature_extraction import DictVectorizer

dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)
```

```python
# fit_transform consists of 2 steps fit and tranform
# fit learns the list of categories for each column like gender has male and female
# transform turns dictionaries into a matrix 
# By default Dictvectorizer returns a sparse matrix, a data structure which stores
# only non zero values coz most of the matriz is zeroes. They are more memory
# efficient but for this dataset it is simpler to work with a dense array

# after fit_tranfrom dv has the fit which means it has learned the rules or
# how many categories are there for each columm which we can use for test
# and validation set later on so no need to perform fit there only transform
```

```python
# To see what columns the vectorizer created, use dv.get_feature_names_out(). 
# For the full dataset we get 45 names: one column per 
# category value (contract=month-to-month, internetservice=fiber_optic, ...) 
# plus the three numerical columns. The resulting matrix has 45 columns,
# one per feature name

X_train.shape
```

```python
# 4225 is the number of rows in training set, 45 is number of features after 
# encoding
```

```python
# Encoding the validation set
# The validation set must be encoded in exactly the same way as the training set.
# Thats why we call only tranform on the validaton set and use the already
# fitted vectorizer 
val_dict = df_val[categorical + numerical].to_dict(orient = 'records')
X_val = dv.transform(val_dict)
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
sigmoid(-100),sigmoid(4), sigmoid(100)
# as we can see from values below for large negative numbers it tends to 0
# ans for large positive numbers it tenda to 1
```

```python
w0 = 2
w = np.array([1.4,2.9,4.7])
def logistic_regresssion(xi):
    score = w0
    
    score = score + xi.dot(w)
    result = sigmoid(score)
    return result
```

# Training logistic regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
```

```python
model.intercept_[0]
```

```python
model.coef_[0].round(3)
```

```python
y_pred = model.predict_proba(X_val)[:,1]
y_pred
```

```python
churn_decision = (y_pred>=0.5)
```

```python
(y_val == churn_decision).mean()
```

```python
# 80% accuracy achieved
```
